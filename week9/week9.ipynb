```python
# NAME: rentalcar.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

# define daily rates for each car type
def get_daily_rate(code):
    if code == "B":
        return 40
    elif code == "S":
        return 60
    elif code == "L":
        return 80


# define full classification name
def get_classification_name(code):
    if code == "B":
        return "Budget"
    elif code == "S":
        return "Standard"
    elif code == "L":
        return "Luxury"


# define free mileage included and charge per mile after initial
def charged_miles(total_miles):
    if total_miles > 100:
        return total_miles - 100
    else:
        return 0


# fruitful function to calculate total bill
def calculate_total_bill(code, days_rented, total_miles):
    daily_rate = get_daily_rate(code)
    extra_miles = charged_miles(total_miles)

    day_cost = daily_rate * days_rented
    mileage_cost = extra_miles * 0.35

    total_bill = day_cost + mileage_cost
    return total_bill


# PROGRAM LOGIC

code = input("Enter classification code (B, S, L): ")

# make it not caps sensitive
code = code.upper()

# validate classification code
while code != "B" and code != "S" and code != "L":
    print("Invalid code. Please enter B, S, or L.")
    code = input("Enter classification code (B, S, L): ")
    code = code.upper()

# prompt for start odometer
start_odometer = int(input("Start Odometer: "))

# prompt for end odometer
end_odometer = int(input("End Odometer: "))

# prompt for days rented
days_rented = int(input("Days Rented: "))

# calculate total miles
total_miles = end_odometer - start_odometer

# calculate charged miles
extra_miles = charged_miles(total_miles)

# call bill function and store returned total
total_bill = calculate_total_bill(code, days_rented, total_miles)

# get full classification name
classification_name = get_classification_name(code)

# create rental summary based on user inputs
print("--- RENTAL SUMMARY ---")
print("Classification:", classification_name)
print("Total Miles:", total_miles)
print("Charged Miles:", extra_miles)
print("Total Bill: $" + str(total_bill))
```
