```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def add_students(book):
    while True:
        student_name = input("Enter student name (or 'done'): ")

        if student_name == "done":
            break

        try:
            student_score = float(input(f"Enter {student_name}'s grade (0-100): "))

            if 0 <= student_score <= 100:
                book[student_name] = student_score
            else:
                print("Grade must be between 0 and 100.")

        except ValueError:
            print("Please enter a valid number.")

# PROGRAM LOGIC

def main():
    gradebook = {}

    add_students(gradebook)

    # Test output to verify dictionary
    print("\nCurrent Gradebook:")
    for name, score in gradebook.items():
        print(f"{name}: {score}")

main()
```
