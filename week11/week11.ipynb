```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def add_students(book):
    while True:
        student_name = input("Enter student name (or 'done'): ")

        if student_name == "done":
            break

        try:
            student_score = float(input(f"Enter {student_name}'s grade (0-100): "))

            if 0 <= student_score <= 100:
                book[student_name] = student_score
            else:
                print("Grade must be between 0 and 100.")

        except ValueError:
            print("Please enter a valid number.")


def calculate_average(book):
    if len(book) == 0:
        return 0

    total = sum(book.values())
    average = total / len(book)

    return average


# PROGRAM LOGIC

def main():
    gradebook = {}

    add_students(gradebook)

    print("\n--- Class Report ---")

    if len(gradebook) == 0:
        print("No students entered.")
    else:
        for name, score in gradebook.items():
            print(f"{name}: {score}")

    class_avg = calculate_average(gradebook)
    print(f"Class Average: {class_avg:.2f}")


main()
```
