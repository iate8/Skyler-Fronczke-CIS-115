```python
# IMPORT STATEMENTS
# N/A

# FUNCTION DEFINITIONS
def calculate_rectangle(length, width):

# Calculate the area of the rectangle.
    area = length * width

# Calculate the perimeter of the rectangle.
    perimeter = 2 * (length + width)

# Return both the calculated area and perimeter as a tuple.
    return (area, perimeter)


# PROGRAM LOGIC

# Ask the user to input the length of the rectangle
# Convert the input string to a float
user_length_str = input("Enter length: ")
user_length = float(user_length_str)


# Ask the user to input the width of the rectangle.
# Convert the input string to a float
user_width_str = input("Enter width: ")
user_width = float(user_width_str)


# Call the calculate_rectangle function with the user's input.
# The function returns a tuple (area, perimeter).
results_tuple = calculate_rectangle(user_length, user_width)


my_area = results_tuple[0]
my_perimeter = results_tuple[1]


# Print the result
print("\n--- Results ---")
print(f"The Area is: {my_area}")
print(f"The Perimeter is: {my_perimeter}")

```
