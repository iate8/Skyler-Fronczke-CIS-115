```python
import datetime
import json
import os

# CLASS DEFINITIONS
class Task:
    def __init__(self, name, priority):
        self.name = name
        self.priority = priority
        self.timestamp = datetime.datetime.now().isoformat() # Store as ISO format string

    def to_dict(self):
        return {
            'name': self.name,
            'priority': self.priority,
            'timestamp': self.timestamp
        }

    def __str__(self):
        return f"Task: {self.name}, Priority: {self.priority}, Created: {self.timestamp.split('T')[0]}"


class TaskManager:
    def __init__(self, active_tasks_file='active_tasks_commit1.json', completed_tasks_file='completed_tasks_commit2.json'):
        self.active_tasks = []
        self.completed_tasks = []
        self.active_tasks_file = active_tasks_file
        self.completed_tasks_file = completed_tasks_file
        self._load_tasks_from_file()
        self._load_completed_tasks_from_file()

    def _load_tasks_from_file(self):
        if os.path.exists(self.active_tasks_file):
            with open(self.active_tasks_file, 'r') as f:
                tasks_data = json.load(f)
                self.active_tasks = []
                for data in tasks_data:
                    task = Task(data['name'], data['priority'])
                    task.timestamp = data['timestamp']
                    self.active_tasks.append(task)

    def _save_tasks_to_file(self):
        with open(self.active_tasks_file, 'w') as f:
            json.dump([task.to_dict() for task in self.active_tasks], f, indent=4)

    def _load_completed_tasks_from_file(self):
        if os.path.exists(self.completed_tasks_file):
            with open(self.completed_tasks_file, 'r') as f:
                tasks_data = json.load(f)
                self.completed_tasks = []
                for data in tasks_data:
                    task = Task(data['name'], data['priority'])
                    task.timestamp = data['timestamp']
                    self.completed_tasks.append(task)

    def _save_completed_tasks_to_file(self):
        with open(self.completed_tasks_file, 'w') as f:
            json.dump([task.to_dict() for task in self.completed_tasks], f, indent=4)

    def add_task(self, name, priority):
        new_task = Task(name, priority)
        self.active_tasks.append(new_task)
        self._save_tasks_to_file()
        print(f"Task '{name}' added with priority {priority}.")

    def complete_task(self, task_index):
        if 0 <= task_index < len(self.active_tasks):
            completed_task = self.active_tasks.pop(task_index)
            self.completed_tasks.append(completed_task)
            self._save_tasks_to_file()
            self._save_completed_tasks_to_file()
            print(f"Task '{completed_task.name}' marked as completed.")
        else:
            print("Invalid task number.")

    def view_tasks(self):
        print("\n--- Active Tasks ---")
        if not self.active_tasks:
            print("No active tasks.")
        else:
            for i, task in enumerate(self.active_tasks):
                print(f"{i+1}. {task}")

    def view_completed_tasks(self):
        print("\n--- Completed Tasks ---")
        if not self.completed_tasks:
            print("No completed tasks.")
        else:
            for i, task in enumerate(self.completed_tasks):
                print(f"{i+1}. {task}")

# PROGRAM LOGIC (AI_BASE_BUILD)
# Create an instance of TaskManager
task_manager_commit1 = TaskManager()

def display_menu_commit1():
    print("\n--- Personal Task Manager (Commit 2) ---")
    print("1. Add New Task")
    print("2. View Active Tasks")
    print("3. Complete a Task")
    print("4. View Completed Tasks")
    print("5. Exit")

def main_program_commit1():
    while True:
        display_menu_commit1()
        choice = input("Enter your choice: ")

        if choice == '1':
            name = input("Enter task name: ")
            while True:
                try:
                    priority = int(input("Enter priority (1: High, 2: Medium, 3: Low): "))
                    if priority in [1, 2, 3]:
                        break
                    else:
                        print("Priority must be 1, 2, or 3.")
                except ValueError:
                    print("Invalid input. Please enter a number.")
            task_manager_commit1.add_task(name, priority)
        elif choice == '2':
            task_manager_commit1.view_tasks()
        elif choice == '3':
            task_manager_commit1.view_tasks()
            if task_manager_commit1.active_tasks:
                try:
                    task_num = int(input("Enter the number of the task to complete: "))
                    task_manager_commit1.complete_task(task_num - 1)
                except ValueError:
                    print("Invalid input. Please enter a number.")
            else:
                print("No active tasks to complete.")
        elif choice == '4':
            task_manager_commit1.view_completed_tasks()
        elif choice == '5':
            print("Exiting Task Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

# Run the main program logic for Commit 1
# main_program_commit1() # Uncomment to run this version
```
