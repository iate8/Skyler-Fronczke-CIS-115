```python
# IMPORT STATEMENTS
#none

# FUNCTION DEFINITIONS

# define is_palindrome
def is_palindrome(word):

# make it case insensitive, reverse using string slicing, and compare to initial word
    word_lower = word.lower()
    reversed_word = word_lower[::-1]
    return word_lower == reversed_word

# PROGRAM LOGIC
# prompt user for a word
user_word = input("Enter a word: ")


#call function
#print "true" or "false"

if is_palindrome(user_word):
    print(f'\'{user_word}\' is a palindrome.')
else:
    print(f'\'{user_word}\' is not a palindrome.')

```
