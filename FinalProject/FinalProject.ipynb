```python
# IMPORT STATEMENTS
# N/a

# FUNCTION DEFINITIONS

def add_expense(expenses_list, category, vendor, amount, date):
    record = {
        'category': category.lower(),
        'vendor': vendor.lower(),
        'amount': amount,
        'date': date
    }
    expenses_list.append(record)

def get_total(expenses_list, filter_key=None, filter_val=None):
    total = 0.0
    for item in expenses_list:
        if filter_key is None:
            total += item['amount']
        elif item[filter_key] == filter_val.lower():
            total += item['amount']
    return total

# main
expenses = []

while True:
    print("\n--- Expense Tracker ---")
    print("1: Add New Expense")
    print("2: Get Totals")
    print("3: Exit")

    choice = input("Selection: ").strip()

    if choice == '3':
        break
    elif choice == '1':
        cat = input("Enter Category: ")
        ven = input("Enter Vendor: ")
        try:
            amt = float(input("Enter Amount: "))
        except ValueError:
            print("Error: Amount must be a number.")
            continue
        dat = input("Enter Date: ")
        add_expense(expenses, cat, ven, amt, dat)
    elif choice == '2':
        print("\n--- Totaling Options ---\nA: All\nC: Category\nV: Vendor")
        sub = input("Choice: ").strip().upper()
        if sub == 'A':
            res = get_total(expenses)
            print(f"Total: ${res:.2f}")
        elif sub == 'C':
            val = input("Category: ")
            res = get_total(expenses, 'category', val)
            print(f"Total for {val}: ${res:.2f}")
        elif sub == 'V':
            val = input("Vendor: ")
            res = get_total(expenses, 'vendor', val)
            print(f"Total for {val}: ${res:.2f}")
```
