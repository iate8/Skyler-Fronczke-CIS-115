```python
# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS

# 1. Define verify_pin():
# - Create a loop that asks for a PIN until it equals "1234"
# - Return True

def verify_pin():
  while True:
    pin = input("Enter PIN: ")

    if pin == "1234":
        print("PIN Verified.")
        return True
  else:
    print("Incorrect PIN. Try again.")


# PROGRAM LOGIC

# Call verify_pin() to test it
verify_pin()
```
