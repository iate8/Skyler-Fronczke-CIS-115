```python
# IMPORT STATEMENTS
# N/A


# FUNCTION DEFINITIONS
def write_entry(filename, entry_text):
    # Open the file in append mode ('a')
    with open(filename, 'a') as file:
        # Write the entry to the file, ensuring each new entry starts on a new line.
        file.write(entry_text + '\n')
    # File is automatically closed by the 'with' statement


def read_journal(filename):
    try:
        # Open the file in read mode ('r')
        with open(filename, 'r') as file:
            print("\n--- Your Journal ---")
            # Read the entire contents and print them to the screen
            content = file.read()
            if not content.strip(): # Check if content is empty or just whitespace
                print("Your journal is empty. Write an entry first!")
            else:
                print(content.strip())
            print("--------------------\n")
        # File is automatically closed by the 'with' statement
    except FileNotFoundError:
        # Handle the error that occurs if the file doesn't exist yet
        print(f"\n--- Your Journal ---")
        print(f"No journal entries found. Start by writing your first entry!")
        print("--------------------\n")


# PROGRAM LOGIC
JOURNAL_FILENAME = 'journal.txt'


print("Welcome to your Personal Journal App!")


while True:
    choice = input("(W)rite, (R)ead, or (Q)uit? ").strip().lower()


    if choice == 'q':
        print("Exiting Journal App. Goodbye!")
        break
    elif choice == 'w':
        entry = input("Enter your entry: ")
        write_entry(JOURNAL_FILENAME, entry)
        print("Entry saved.\n")
    elif choice == 'r':
        read_journal(JOURNAL_FILENAME)
    else:
        print("Invalid choice. Please enter 'W', 'R', or 'Q'.\n")

```
