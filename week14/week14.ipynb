```python
# IMPORT STATEMENTS
# N/A

# CLASS DEFINITIONS
# Define a class named 'Book'
class Book:
    # Define the '__init__' method that accepts 'self', 'title', and 'author'
    def __init__(self, title, author):
        # Assign 'title' to self.title
        self.title = title
        # Assign 'author' to self.author
        self.author = author
        # Initialize 'is_available' to True for self.is_available
        self.is_available = True

    # Define the '__str__' method that accepts 'self'
    def __str__(self):
        # If self.is_available is True:
        #   Return a string like "[Title] by [Author] - Available"
        # Else (if self.is_available is False):
        #   Return a string like "[Title] by [Author] - Checked Out"
        status = "Available" if self.is_available else "Checked Out"
        return f"{self.title} by {self.author} - {status}"

    # Define the 'checkout' method that accepts 'self'
    def checkout(self):
        # Set self.is_available to False
        self.is_available = False

    # Define the 'return_book' method that accepts 'self'
    def return_book(self):
        # Set self.is_available to True
        self.is_available = True

# PROGRAM LOGIC
# Create at least two instances (objects) of the Book class
book1 = Book("Think Python", "Allen Downey")
book2 = Book("The Hobbit", "J.R.R. Tolkien")

print("--- Library Status ---")
# Print the status of both book objects
print(book1)
print(book2)

print("\nChecking out 'The Hobbit'...")
# Call the 'checkout' method on one of the book objects
book2.checkout()

print("\n--- Updated Status ---")
# Print the status of both book objects again to show the change
print(book1)
print(book2)

print("\nReturning 'The Hobbit'...")
book2.return_book()

print("\n--- Final Status ---")
print(book1)
print(book2)
```
